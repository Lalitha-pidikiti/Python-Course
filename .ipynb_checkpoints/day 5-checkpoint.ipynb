```python
s1 = {2,1,4,6,5}
s2 = {1,8,7,8}
print(s1.union(s2))
print(s1&s2)
```

```python
s = {2,1,4,6,5}
s1 = {1,4,6}
print(s|s1)
print(s-s1)
print(s^s1)
print(s.difference(s1))
print(s.symmetric_difference(s1))
```
