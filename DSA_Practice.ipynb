```python
n = int(input())
rev = 0
while n>0:
    digit = n%10
    rev = rev*10+digit
    n=n//10
print(rev)
```

```python
# prime or not 
for n in range(2 ,101):
    count = 0
    for i in range(1,n+1):
        if n%i ==0:
            count +=1
    if count==2:
        print(n)
    
```

```python
l = [1,2,3,4,5]
even_indexes = []
for index in range(len(l)):
    if l[index]%2==0:
        even_indexes.append(index)
```

```python
sentence = "Python code code"
words = sentence.split()
f = {}
for word in words:
    if word in f:
        f[word]=f[word]+1
    else:
        f[word] = 1
print(f)
#print(words)
```

```python
sentence="Python code code"
words=sentence.split()
count={}
for word in words:
    if word in count:
        count[word] += 1
    else:
        count[word]=1
most_repeated=max(count,key=count.get)
print("Most repeated word:",most_repeated)
print("Count:",count[most_repeated])
```
