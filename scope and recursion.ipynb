```python
#1. local scope
def f():
    x = 10
    print(x)
f()
```

```python
#2. Gobal scope
x = 10
def f1():
    print(x)
f1()
print(x)
```

```python
#3. enclosing scope
def outer():
    x = "outer"
   # print(x)
    def inner():
        y = "inner"
        #print(y)
        print(x,y)
    inner()
outer()
```

```python
#4.Built in scope
def f3():
    l = [1,2,3]
    return sum(l)
f3()
```

```python
#  reccursion
def factorial(n):
    if n == 1 or n == 0:
        return 1
    return n*factorial(n-1)
factorial(5)
```

```python
def sum_of_natural_numbers(n):
    if n==1:
        return 1
    return n+sum_of_natural_numbers(n-1)
print(sum_of_natural_numbers(10))
```

```python
def sum_of_natural_numbers(n):
    if n==2:
        return 2
    if n==1:
        return 1
    return n+sum_of_natural_numbers(n-2)
print(sum_of_natural_numbers(9))
```

```python
def sum_of_natural_numbers(n):
    if n%2==0:
        if n==2:
            return 2
        return n+sum_of_natural_numbers(n-2)
    else:
        return sum_of_natural_numbers(n+1)
print(sum_of_natural_numbers(9))
```

```python
def prime_number(n):
    count=0
    for i in range(1,n+1):
        if n%i==0:
            count +=1
    if count == 2:
        return True
    else:
        return False
num = int(input())
print(prime_number(num))
        
```

```python
def pnc(n):
    count=0
    for i in range(1,n+1):
        if n%i==0:
            count +=1
    if count == 2:
        return True
    else:
        return False
def pn_list(num):
    l = []
    for number in range(2,num+1):
        if pnc(number):
            l.append(number)
    return 1
num = int(input())    
pn_list(num)       
```
