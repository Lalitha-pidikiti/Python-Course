```python
#1.with return & with parameters 
def division(a,b):
    return a%b
print(division(10,2))
```

```python
#2.with return & without parameters
def division1():
    a=6
    b=int(input())
    return a%b
z = division1()
print(z)
```

```python
#3.with out return &with parameters
def division2(x,y):
    print(x%y)
a = division2(16,8)
print(a)
```

```python
#4.with out return &with out parameters
def division3():
    a=4
    b=8
    print(a%b)
a=division3()
print(a)
```

```python
#positional arguments
def addition(a,b):
    return a+b
print(addition(10,20))
```

```python
#keword arguments
def addition(a,b):
    print(a,b)
    return a+b
c = 45
d = 55
print(addition(b=c,a=d))
    
```

```python
#default arguments
def addition1(a,b,c=0,d=0):
    return a+b+c+d
print(addition1(10,20))
print(addition1(10,20,30))
print(addition1(10,20,30,40))
```

```python
#variable length arguments
# variable length follow avvali ante *ni mention cheskovali
def addition5(*var):
    print(type(var))
    #return sum(var)
    total = 0
    for num in var:
        total +=num
    return total
print(addition5(10,20,30))
print(addition5(1,3,2,4,5,6,7,8,9))
```

```python
# variable length keyword argument 
def addition5(**var):
    print(type(var))
    #return sum(var)
    total = 0
    for num in var.values():
        total +=num
    return total
#print(addition5(10,20,30))
print(addition5(a=1,b=2,c=3))
```

```python
# variable length keyword argument 
def addition5(**var):
    print(type(var))
    #return sum(var)
    total = ""
    for num in var.keys():
        total +=num
    return total
#print(addition5(10,20,30))
print(addition5(a=1,b=2,c=3))
```
