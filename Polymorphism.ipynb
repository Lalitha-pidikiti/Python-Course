```python
class calculation:
    def addition(self,a,b):
        return a+b
    def addition(self,a,b,c):
        return a+b+c
a = calculation()
a.addition(10,20)
```

```python
class cal:
    def addition1(self,a,b,c=0,d=0):
        return a+b+c+d
b = cal()
print(b.addition1(10,20))
print(b.addition1(10,20,5))
print(b.addition1(1,2,3,4))
```

```python
# method overload
class cal:
    def addition1(self ,*a):
        return sum(a)
b = cal()
print(b.addition1(10,20))
print(b.addition1(10,20,5))
print(b.addition1(1,2,3,4))
print(b.addition1(1,2,3,4,5,6,8,7,9))
```

```python
#method overriding
class A:
    def m1(self):
        print("parent m1")
class B(A):
    def m1(self):
        print("child m1")
        super().m1()
c1 = B()
c1.m1()
A.m1(c1)
```

```python
#method overriding
class Employee:
    def work(self):
        #print("Employee is working")
        return "Employee is working"
class Developer(Employee):
    def work(self):
        #super().work()
        #print("Developer is working")
        return "Developer is Working"
#d = Developer()
#d.work()
#Employee.work(d)
res = [Employee(),Developer()]
for i in res:
    print(i.work())
```
